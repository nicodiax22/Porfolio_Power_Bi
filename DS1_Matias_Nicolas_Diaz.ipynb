#Realizamos las primeras modificaciones y definimos las primeras agrupaciones ✍

1.   Eliminamosc columnas que no sirven para el analisis
2.   Cambiamos el tipo de dato a las columnas de fecha

```python
#Eliminamos las columnas donde los valores sean igual a la cantidad de lineas del dataframe
#Eliminamos las columnas donde los valores sean igual a 1
#Eliminamos las filas duplicadas
#Estos datos no nos sirven para realizar analisis

df.drop(df.columns[df.nunique() == df.shape[0]], axis=1, inplace=True)
df.drop(df.columns[df.nunique() == 1], axis=1, inplace=True)
df = df.drop_duplicates()
```

```python
#Cambiamos los datos de fecha de objeto a datetime

df['Order Date']= pd.to_datetime(df['Order Date'])
df['Ship Date']= pd.to_datetime(df['Ship Date'])

# Separamos la columna 'Ship Date' para obtener columnas de año, mes y dia

df['Year'] = df['Ship Date'].dt.year
df['Month'] = df['Ship Date'].dt.month
df['Day'] = df['Ship Date'].dt.day
```

```python
#Pasamos la columna Order Data al indice
df.set_index('Order Date', inplace=True)
```

```python
#Vamos a agregar columna de costo

df['Cost']= df['Sales']- df['Profit']
df['Cost']
```

```python
#Agrupamos las ventas por año y por region

salesXyear= df.groupby('Region')['Sales'].resample('ME').sum().reset_index()
salesXyear['Sales']= salesXyear['Sales'].astype(int)
salesXyear = salesXyear.rename(columns={'Order Date': 'Month and year'})
print(salesXyear)
```

```python
#Agrupamos las Cantidades vendidas por año y por region

qtyXyear= df.groupby('Region')['Quantity'].resample('ME').sum().reset_index()
qtyXyear = qtyXyear.rename(columns={'Order Date': 'Month and year'})
print(qtyXyear)
```

```python
#Agrupamos las ventas por sub-category y el año
#Cambiamos tipo de dato de columna sales
#Ordenamos de menor a mayor las ventas e imprimimos

subcategoryXsales= df.groupby('Sub-Category')['Sales'].resample('ME').sum().reset_index()
subcategoryXsales['Sales']= subcategoryXsales['Sales'].astype(int)
subcategoryXsales = subcategoryXsales.rename(columns={'Order Date': 'Month and year'})

print(subcategoryXsales)
```

# **Tecnica de deteccion de Outliers: Rango Intercuartil**

```python
# Calcular Q1 (primer cuartil) y Q3 (tercer cuartil)

Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)

# Calcular el rango intercuartil (IQR)
IQR = Q3 - Q1

# Definir los limites para detectar Outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identificar los Outliers
outliers = df[(df['Sales'] < lower_bound) | (df['Sales'] > upper_bound)]

outliers
```

# **Tecnica de deteccion de Outliers: Z-Score**

```python
from scipy import stats

# Calcular el z-score para cada valor en la columna 'Sales'
df['z_score'] = stats.zscore(df['Sales'])

# Filtrar Outliers usando el umbral de Z-Score
outliers_zscore = df[df['z_score'].abs() > 1.5 ]
print("Outliers detectados con zscore")
outliers_zscore
```

# **Acciones sobre los Outliers**

**Eliminar Outliers del dataframe original usando el rango intercuartil**

```python
# Eliminar Outliers del dataframe original usando el rango intercuartil

df_cleaned_RQ = df[~((df['Sales'] < lower_bound) | (df['Sales'] > upper_bound))]
df_cleaned_RQ
```

**Eliminar Outliers del dataframe original usando Z-Score**

```python
# Eliminar Outliers del dataframe original usando Z-Score

df_cleaned_ZS = df[df['z_score'].abs() <= 1.5]
df_cleaned_ZS
```

**En una copia del dataframe original, remplazar Outliers con la media usando Z-Score**

```python
#Creo una copia de df

df_cleaned_md = df.copy()

# Remplazar los Outliers con la mediana utilizando la copia del df
df_cleaned_md['Sales']= df_cleaned_md['Sales'].where(df['z_score'].abs() <= 1.5, df['Sales'].median())
df_cleaned_md
```

```python
# Importamos lo necesario para utilizar ARIMA

from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

#Asegurarme que el df esté en el formato correcto

salesXyear= salesXyear.rename(columns={'Month and year': 'ds', 'Sales': 'y'})
salesXyear.set_index('ds', inplace=True)

#Verifica la estacionariedad de la serie

result = adfuller(salesXyear['y'])
print('ADF Statistic: %f' % result[0])
print('p-value: %f' % result[1])


# Determinar los parametros p, d, q

plot_acf(salesXyear['y'])
plot_pacf(salesXyear['y'])

#Crea el modelo ARIMA

model = ARIMA(salesXyear['y'], order=(1, 1, 1))
model_fit = model.fit()


# Haz la prediccion

forecast = model_fit.forecast(steps=3)
print(forecast)
```
